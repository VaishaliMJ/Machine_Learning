```python
from sklearn.datasets import make_regression
```

```python
X,y=make_regression(n_samples=4,n_features=1,n_targets=1,noise=80,random_state=13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(X,y)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
# Regression line
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red')
```

```python
# m=78.75,b=0

y_pred=((78.35*X)+(0)).reshape(4)
```

```python
plt.plot(X,lr.predict(X),color='red',label="OLS")
plt.plot(X,y_pred,color="black",label="b=0")
plt.legend()
plt.show()
```

```python
import matplotlib.animation as animation
import numpy as np


m=78.35
b=-100
lr=0.01
epoch=100

for i in range(epoch):
  updated_slope=-2*np.sum(y-m*X.ravel()-b)
  b=b-(lr*updated_slope)

  y_pred=m*X+b


  plt.plot(X,y_pred)
plt.scatter(X,y)
```
