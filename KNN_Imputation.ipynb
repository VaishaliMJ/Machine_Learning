```python
import pandas as pd
```

```python
df=pd.read_csv("train.csv")

df.head()
```

```python
titanic=df[['Age','Pclass','Fare','Survived']]

titanic.head()
```

```python
titanic.isnull().sum()
```

```python
titanic.isnull().mean()*100
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.impute import SimpleImputer,KNNImputer
```

```python
x=titanic.drop(columns=['Survived'],axis=1)
y=titanic['Survived']
```

```python
xTrain,xTest,yTrain,yTest=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
knnImputor=KNNImputer(n_neighbors=3,weights='uniform')
xTrain_trf=knnImputor.fit_transform(xTrain)
xTest_trf=knnImputor.transform(xTest)
```

```python
pd.DataFrame(xTrain_trf,columns=xTrain.columns)
```

```python
lr=LogisticRegression()
lr.fit(xTrain_trf,yTrain)
yPred=lr.predict(xTest_trf)
accuracy_score(yTest,yPred)
```

```python
simpleImputor=SimpleImputer()
xTrain_trf2=simpleImputor.fit_transform(xTrain)
xTest_trf2=simpleImputor.transform(xTest)

lr=LogisticRegression()
lr.fit(xTrain_trf2,yTrain)
yPred=lr.predict(xTest_trf2)
accuracy_score(yTest,yPred)
```
